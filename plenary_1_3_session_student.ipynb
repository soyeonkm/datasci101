# Round 1: Collections and indexing

A **collection** holds more than one thing. An **ordered** collection
organizes its contents by **position**, so we can ask for the item at a
given position.

```python
lst = [11, 22, 33]        # a list: any types, any length, can be changed
mystr = "Data Science"    # a string is a collection of characters
tup = (42.5, -8)          # a tuple: fixed size, cannot be changed
DS101 = {"dept": "Data Science", "number": 101}   # a dictionary: keys to values
```

**Indexing** uses square brackets and starts at zero. Negative positions
count from the end.

```python
lst[0]
```

A **slice** takes a range of positions, up to but not including the
second.

```python
mystr[0:4]
```

A dictionary is **unordered** – there is no “first” item – so we index
it by **key** instead of by position.

```python
DS101["dept"]
```

`len` gives the size of any of them, and `in` searches.

```python
33 in lst
```

# Round 3: List comprehensions

Doing the same thing to every item is so common that Python has a
compact way to write it. Read it right to left: start with
`temperatures`, let `t` be each item, and collect the result of the
computation.

```python
[t + 1 for t in temperatures]
```

Adding `if` at the end keeps only some of the items.

```python
[t for t in temperatures if t > 70]
```

# Round 4: Conditionals

An `if` statement picks which block of code runs, `elif` adds further
checks, and `else` catches everything left. Only the **first** matching
branch runs, which is what makes the ordering matter.

```python
def describe(temp):
    if temp < 60:
      return "cold"
    elif temp < 80:
      return "warm"
    else:
      return "hot"

[describe(t) for t in temperatures]
```

The `elif` does not need to say `temp >= 60 and temp < 80`. Getting
there at all means the first test already failed.
