### The rules we will use

- Roll **four dice**, split them into **two pairs**, and read the sum of
  each pair. Those sums name **columns** (2 to 12).
- Advance one space in each column named, but you may only ever have
  **three columns** going at once.
- You **must place every marker you can**. You may not decline a move to
  stay safe.
- Then choose: **stop** and bank your progress, or **roll again**. If a
  new roll gives you nothing you can use, you **bust** and lose
  everything from that turn.

Both are random because the roll is; the positions are not — you chose
to stand there. Note that $X = 0$ is a bust.

### A note on sets

Plenary 1.3 introduced **sets** — unordered collections, written
`{1, 2, 3}`, that never hold the same thing twice. That is exactly what
we need for “the columns a roll can reach”, so here is everything we
will do with one.

```python
columns = set()          # an empty set -- note set(), not {}, which is a dictionary
columns.add(7)           # .add puts something in
columns.add(7)           # again: no effect, a set never holds a repeat
columns.add(4)
{"the set": columns,
 "is 7 in it": 7 in columns,
 "how big": len(columns)}
```

The one operator we need is `&`, which gives what is in **both** sets.
Put `len` around it and you have counted the overlap — which is exactly
what $X$ and $Y$ are.

```python
{"in both": {6, 7, 8} & {7, 8, 9},
 "how many of 6,7,8 are in {4,7}": len({6, 7, 8} & {4, 7}),
 "how many of 6,7,8 are in {2,3}": len({6, 7, 8} & {2, 3})}
```
