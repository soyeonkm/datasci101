## Question 2: The sums you could make

Four dice split into two pairs in exactly **three** ways, and each way
gives you **two sums** — the two columns you could move on.

Write:

- **`possible_sums(roll)`**
- `roll` — a list of four dice
- returns a **list of three tuples**, each holding the two sums

Take the three ways **in this order**, so that your answers match the
checks: first die with second, first with third, first with fourth. (In
each case the two dice left over form the other pair.) Unpack the roll
with `a, b, c, d = roll` and the three ways are three bits of

Demonstrate on `[1, 2, 3, 4]` and `[1, 1, 2, 4]`. In the first, one of

```python
# your code here
def possible_sums(roll):
    a, b, c, d = roll
    return [
        (a + b, c + d),
        (a + c, b + d),
        (a + d, b + c),
    ]

print(possible_sums([1, 2, 3, 4]))
print(possible_sums([1, 1, 2, 4]))

```

## Question 3: Can you use a sum?

Whether you may move on a column comes down to **how many markers you
have down already**, and there are three cases:

- the column is **already yours** — always fine, you just move the
  marker you have there
- it is a **new** column and you have **fewer than three** markers down
  — fine, you place a new one
- it is a **new** column and **all three** markers are down — you cannot

Write:

- **`can_place(total, active)`**
- `total` — one column number
- `active` — the set of columns you currently hold markers on
- returns **`True` or `False`**

The first two cases have the same answer, so you do not need three
branches. Write this as a **single `return`**: `in` for the first test,
`len` for the second, `or` to join them. The condition *is* the answer,
so there is nothing to branch on.

Then show all three cases: column 7 while holding `{6, 7, 8}` (already
yours), column 2 while holding `{6, 7}` (room to spare), and column 2
while holding `{6, 7, 8}` (no room).

```python
# your code here
def can_place(total, active):
    return total in active or len(active) < 3

print(can_place(7, {6, 7, 8}))
print(can_place(2, {6, 7}))
print(can_place(2, {6, 7, 8}))
```

## Question 4: Busting

You **bust** when a roll gives you nothing you can use: no way of
pairing the dice offers a single sum you can place.

Rather than answer that directly, first gather up everything you *could*
do.

Write:

- **`usable_sums(roll, active)`**
- `roll` — a list of four dice
- `active` — the set of columns you hold
- returns a **list of every sum you could legally move on**, across all
  three ways of pairing the dice. The same sum can appear more than once
  — that is fine, we only care whether the list is empty.

Build it the way plenary 1.3 collects results: start with an empty list,
name it, loop over the three ways and over the two sums in each, and
`.append` any sum that `can_place` allows.

Then write:

- **`has_busted(roll, active)`**
- same arguments
- returns **`True` if the roll is useless to you**, `False` otherwise

That one is a single line: you have busted exactly when `usable_sums`
comes back empty.

Demonstrate both against `{6, 7, 8}` on `[1, 1, 2, 2]` and on
`[1, 2, 3, 4]`, then `[1, 1, 2, 2]` again against `{6, 7}`. Work each
one out by hand first: write out the three pairings and check them
against the columns you hold. The second roll is less obvious than it
looks.

```python
# your code here
def usable_sums(roll, active):
  results = []
  for way in possible_sums(roll):
      for total in way:
          if can_place(total, active):
              results.append(total)
  return results

def has_busted(roll, active):
    return usable_sums(roll, active) == []

print(sorted(usable_sums([1, 2, 3, 4], {6, 7, 8})))
print(usable_sums([1, 1, 2, 2], {6, 7, 8}))
print(usable_sums([1, 1, 2, 2], {6, 7}))

print(has_busted([1, 1, 2, 2], {6, 7, 8}))
print(has_busted([1, 2, 3, 4], {6, 7, 8}))
print(has_busted([1, 1, 2, 2], {6, 7}))

```

# Part 2: Probability

```python
all_rolls = list(product(range(1, 7), repeat=4))
len(all_rolls)
```

## Question 5: The chance of each column

The eleven columns are **not** equally likely, and that is the whole
reason the board is shaped the way it is. Each column is named by the
sum of **two** dice, and two dice have 36 equally likely outcomes — but
those 36 spread unevenly across the eleven sums.

Write:

- **`pair_chances()`**
- takes **no arguments**
- returns a **dictionary** mapping each sum from 2 to 12 to its
  **probability**

Count with two nested loops over `range(1, 7)`. Start the dictionary
with a **zero for every sum from 2 to 12**, so there is always a count
already there to add one to, then divide each count by 36 at the end.

Add up all eleven probabilities as a check. A pair of dice gives exactly
one sum, so the eleven events overlap nowhere and cover everything that
can happen — they must total exactly 1.

```python
# your code here
def pair_chances():
    counts = {}
    for total in range(2, 13):
        counts[total] = 0

    for die1 in range(1, 7):
        for die2 in range(1, 7):
            counts[die1 + die2] += 1

    result = {}
    for total, count in counts.items():
        result[total] = count / 36

    return result

print(pair_chances())
print(sum(pair_chances().values()))
```

```python
# Check your work. Delete the '# ' from the start of each line below,
# then run this cell. Every check should pass without printing anything.
# An AssertionError means that check found a bug.
assert pair_chances()[7] == 6 / 36
assert pair_chances()[2] == 1 / 36
assert pair_chances()[4] == pair_chances()[10]
assert len(pair_chances()) == 11
assert abs(sum(pair_chances().values()) - 1.0) < 0.000001
```

## Question 6: The chance of busting

**(a)** Now the number a player actually wants: you hold three columns
and are deciding whether to roll again.

Counting the rolls that leave you *something* to do is awkward — six
sums to check, and they overlap, so you cannot add up chances the way
you just did. Counting the rolls that leave you **nothing** is easy: one
condition, which `usable_sums` already tests. So count the busts, then
use the **complement rule** — anything that is not a bust is a roll you
can use, so the two chances add to 1.

Write:

- **`bust_chance(active)`**
- `active` — the set of columns you hold
- returns the **probability** that the next roll busts you

Then show it beside its complement for four positions: `{7}`, `{2, 12}`,
`{6, 7, 8}` and `{2, 3, 12}`.

```python
# your code here
def bust_chance(active):
    busts = 0
    for roll in all_rolls:
        if has_busted(list(roll), active):
            busts = busts + 1
    return busts / len(all_rolls)

for active in [{7}, {2, 12}, {6, 7, 8}, {2, 3, 12}]:
    chance = bust_chance(active)
    print(active, "bust:", chance, " not-bust:", 1 - chance)
```

```python
# Check your work. Delete the '# ' from the start of each line below,
# then run this cell. Every check should pass without printing anything.
# An AssertionError means that check found a bug.
assert bust_chance({7}) == 0
assert bust_chance({2, 12}) == 0
assert abs(bust_chance({6, 7, 8}) - 104 / 1296) < 0.000001
assert abs(bust_chance({2, 3, 12}) - 728 / 1296) < 0.000001
```

This one really is an `if` / `elif` / `else`: the three branches give

# Part 3: Random variables

**Position one: you hold columns 6, 7 and 8** — the three middle

## Question 7: Working out X and Y

**(a)** Write:

- **`reachable(roll)`**
- `roll` — a list of four dice
- returns the **set** of columns the roll can reach: every sum appearing
  in any of its three pairings

Start from an empty `set()` and `.add` each sum as you meet it. Because
a set never holds a repeat, you do not have to check whether a column is
already there.

Demonstrate on `[3, 4, 2, 5]`, which reaches five different columns, and
on `[1, 1, 6, 6]`, which reaches only three.

```python
# your code here
def reachable(roll):
    columns = set()
    for way in possible_sums(roll):
        for total in way:
            columns.add(total)
    return columns

print(reachable([3, 4, 2, 5]))
print(reachable([1, 1, 6, 6]))
```

```python
# Check your work. Delete the '# ' from the start of each line below,
# then run this cell. Every check should pass without printing anything.
# An AssertionError means that check found a bug.
assert reachable([1, 2, 3, 4]) == {3, 4, 5, 6, 7}
assert reachable([1, 1, 1, 1]) == {2}
assert reachable([1, 1, 6, 6]) == {2, 7, 12}
```

**(b)** Now the distributions of $X$ and $Y$, in four steps.

1.  Using `all_rolls` and a **list comprehension**, build a list holding
    the reachable set of every roll. Call it **`all_reachable`**.
2.  From that, build a list of the 1,296 values of $X$: for each roll,
    `len(columns & {6, 7, 8})`. Call it **`x_values`**.
3.  Do the same for $Y$, with `{2, 11, 12}`. Call it **`y_values`**.
4.  Both lists are 1,296 numbers long, which tells you nothing as it
    stands. Turn each into a **frequency table**: a dictionary from each
    possible value to how many rolls gave it. Call them **`x_table`**
    and **`y_table`**.

Use those five names: the rest of Part 3 builds on them. Show both
tables.

For step 4, note that $X$ and $Y$ can only be 0, 1, 2 or 3, so start
each table with a zero for all four values and add one as you meet them.
Starting from zeros rather than an empty dictionary also means a value
that never occurs shows up as 0 instead of going missing — which matters
here.

```python
# your code here
all_reachable = [reachable(list(roll)) for roll in all_rolls]

x_values = [len(columns & {6, 7, 8}) for columns in all_reachable]
y_values = [len(columns & {2, 11, 12}) for columns in all_reachable]

x_table = {value: 0 for value in [0, 1, 2, 3]}
for value in x_values:
    x_table[value] = x_table[value] + 1

y_table = {value: 0 for value in [0, 1, 2, 3]}
for value in y_values:
    y_table[value] = y_table[value] + 1

print(x_table)
print(y_table)
```

```python
# Check your work. Delete the '# ' from the start of each line below,
# then run this cell. Every check should pass without printing anything.
# An AssertionError means that check found a bug.
assert len(all_reachable) == 1296
assert len(x_values) == 1296
assert len(y_values) == 1296
assert sorted(x_table) == [0, 1, 2, 3]
assert sorted(y_table) == [0, 1, 2, 3]
assert sum(x_table.values()) == 1296
assert sum(y_table.values()) == 1296
```

**(c)** Divide through by 1,296 to turn each frequency table into a
distribution — call them **`x_chances`** and **`y_chances`**. Show both,
and show $P(X = 0)$ and $P(Y = 0)$ beside the bust chances from Question
6 for `{6, 7, 8}` and `{2, 11, 12}`.

```python
# your code here
x_chances = {}
for value in x_table:
    x_chances[value] = x_table[value] / 1296

y_chances = {}
for value in y_table:
    y_chances[value] = y_table[value] / 1296

print(x_chances)
print(y_chances)

print("P(X=0):", x_chances[0], " bust chance for {6,7,8}:", bust_chance({6, 7, 8}))
print("P(Y=0):", y_chances[0], " bust chance for {2,11,12}:", bust_chance({2, 11, 12}))
```

## Question 8: Comparing X and Y

**(a)** Compute the **mean** and **variance** of each, the way plenary
1.5 did: the list of $P \times \text{value}$ products summed for the
mean, then the list of $(\text{value} - \text{mean})^2 \times P$
products summed for the variance.

No new functions — you have `x_chances` and `y_chances`, so this is four
sums. Call the results **`x_mean`**, **`y_mean`**, **`x_variance`** and
**`y_variance`**, and show all four.

```python
# your code here

x_mean = sum(value * x_chances[value] for value in x_chances)
y_mean = sum(value * y_chances[value] for value in y_chances)

x_variance = sum((value - x_mean) ** 2 * x_chances[value] for value in x_chances)
y_variance = sum((value - y_mean) ** 2 * y_chances[value] for value in y_chances)

print("x_mean:", x_mean, " y_mean:", y_mean)
print("x_variance:", x_variance, " y_variance:", y_variance)
```

**(b)** Two more summaries, both easier from the **lists** of 1,296

First the **median** — the middle value, which plenary 1.5 defines as

Then the probability each variable is **1 or more** — that the roll

```python
# your code here

x_median = np.median(x_values)
y_median = np.median(y_values)

print("x_median:", x_median, " y_median:", y_median)

x_at_least_one = np.mean([value >= 1 for value in x_values])
y_at_least_one = np.mean([value >= 1 for value in y_values])

print("P(X>=1):", x_at_least_one, " P(Y>=1):", y_at_least_one)
```
