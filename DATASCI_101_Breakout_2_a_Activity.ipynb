## Reference and troubleshooting
This reference gives general syntax patterns, not completed task solutions. Replace placeholder names with names from your own work.

| Goal | Pattern |
|---|---|
| Load a table | `pd.read_csv(url)` |
| Dimensions or names | `table.shape`, `table.columns` |
| Preview rows | `table.head(n)`, `table.tail(n)` |
| Select columns | `table[["column_a", "column_b"]]` |
| Select matching rows | `table.loc[condition]` |
| Combine conditions | `(condition_a) & (condition_b)` |
| Count category values | `series.value_counts()` |
| Rank a table | `table.sort_values("column", ascending=False)` |
| Count missing entries | `series.isna().sum()` |
| Count observed entries | `series.count()` |
| Summary within groups | `table.groupby("group")["value"].median()` |

**Common checks**
- `NameError`: complete and run the cell creating the named table. Check spelling.
- `KeyError`: compare your column name with `movies.columns`.
- `SyntaxError`: check quotes, brackets, and parentheses.
- Ambiguous truth value: use `&` or `|` with parentheses around each pandas comparison.
- Unexpected count: check the starting table, inequality direction, and missing-value policy.
- Unexpected ranking: check sort direction and whether sorting precedes row selection.
