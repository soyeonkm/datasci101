# Round 3: Functions

Functions take **arguments** and **return values**. Extra arguments may
be optional, and naming them lets you pass them out of order.

```python
round(1.66667, 2)
```

```python
round(number = 1.66667, ndigits = 3)
```

`def` defines our own function. The body is **indented** and `return`
sends the value back.

```python
def exfun(pos1, pos2, optional = 7):
  return (pos1 + pos2) / optional

exfun(1, 2)
```

```python
exfun(60, 40, optional = 10)
```

# Round 4: Formatted strings (“f-strings”)

This part is new – it was not in the video.

Instead of converting everything to a string by hand, put an `f` in
front of the quote marks and place variable names in curly braces `{}`.

```python
name = "Jordan"
age = 20
f"My name is {name} and I am {age} years old."
```

Notice how `{name}` and `{age}` are replaced by their values. This works
for integers, floats, and boolean values too, with no casting.

You can also skip the `round` function. Here `.1f` means “one digit
after the decimal point”.

```python
score = 95.678
f"Your score is {score:.1f}."
```

Curly braces can hold a computation, not just a variable name:

```python
f"A 10% lower score would be {score * 0.9:.2f}."
```

# Round 5: Booleans

Boolean values are either `True` or `False`, and like other objects they
can be stored in variables.

```python
is_raining = True
type(is_raining)
```

They usually arrive from comparing two things, using `==`, `!=`, `<`,
`>`, `<=`, and `>=`.

```python
age = 18
is_adult = age >= 18
is_adult
```

The operators `and`, `or`, and `not` build bigger conditions from
smaller ones.

```python
has_password = True
is_verified = False
has_password and is_verified
```

```python
not is_verified
```
